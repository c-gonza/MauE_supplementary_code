```python
import re
import pandas as pd
import os
import re
import pandas as pd
import sys
from multiprocessing import Pool, TimeoutError
import time
from pathlib import Path
import csv
```

```python
df_mapped = pd.read_csv(
    r"G:\uniprot_2025_stuff\uniprot_tremble_stuff\species_indexed2.csv"
)
print(df_mapped)
print(df_mapped['species_name'][0])
print(df_mapped['file_path'][0])
```

```python
taxid_names = pd.read_excel(r"C:\Users\bgonzale24\Desktop\Notebooks for PLOS\phylogen_datatables\taxid2speciesname_table.xlsx")
species_names = list(taxid_names['species_name_org'].dropna().unique())
print(len(species_names))
```

```python
def search_name(q_string, text):
    pattern = re.compile(re.escape(q_string) + r"[^//]*")
    return re.finditer(pattern, text)


species_to_paths = (
    df_mapped
    .groupby("species_name")["file_path"]
    .apply(list)
    .to_dict()
)
required_iprs = [
    'IPR033954',
    'IPR036249',
    'IPR006370',
    'IPR026046',
]

dsbD_iprs = [ # this is the same as 13a
    'IPR051790',
    'IPR028250',
]

dsbA_iprs = [
    'IPR001853',
    'IPR023205',
]

def is_complete(checks):
    return all(checks.values())

dicts2 = []
start_time = time.time()

for n, species in enumerate(species_names[15442:20591], start=1): #this little slicing difference is the most important change
    checks = {
        'IPR033954': False,
        'IPR036249': False,
        'IPR006370': False,
        'IPR026046': False,
        'dsbD': False,
        'dsbA': False,
    }

    paths = dict.fromkeys(species_to_paths.get(species, []))
    print(species)
    for path2 in paths:
        with open(path2, "r", encoding="utf-8", errors="replace") as file:
            text = file.read()

        for hit in search_name(species, text):
            hit_text = hit.group()

            for ipr in required_iprs:
                if not checks[ipr] and ipr in hit_text:
                    checks[ipr] = True

            if not checks["dsbD"] and any(ipr in hit_text for ipr in dsbD_iprs):
                checks["dsbD"] = True

            if not checks["dsbA"] and any(ipr in hit_text for ipr in dsbA_iprs):
                checks["dsbA"] = True

            if is_complete(checks):
                break

        if is_complete(checks):
            break

    checks["species"] = species
    dicts2.append(checks)

    if n % 10 == 0:
        elapsed = time.time() - start_time
        print(f"{n} species done | avg {elapsed / n:.2f} sec/species") #same

full_df = pd.DataFrame.from_records(dicts2)
```

```python
full_df = pd.DataFrame.from_records(dicts2)
print(full_df)
full_df.to_excel(r"Path\full_df2.xlsx")
```
